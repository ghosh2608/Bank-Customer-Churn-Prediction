SVM (POLYNOMIAL KERNEL)

```python
#Load libraries
import warnings
warnings.filterwarnings('ignore')
import joblib
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC
from evaluation_script import evaluate_model
```

```python
#Load Processed Data
data = joblib.load('../processed_data/dataset_bundle.pkl')
X_train = data['X_train']
y_train = data['y_train']
X_test = data['X_test']
y_test = data['y_test']
```

```python
#Declare parameters
param_grid = {'C': [0.1, 1, 10, 100],
              'degree': [2, 3, 4],
              'gamma': ['scale', 'auto'],
              'coef0': [0.0, 0.5, 1.0]}
```

```python
#Create Search Space
svm_poly_Grid = GridSearchCV(SVC(kernel='poly', class_weight='balanced', probability=True, random_state=200),
                               param_grid,
                               cv=10,
                               refit=True,
                               scoring='f1',
                               verbose=0)
```

Fit Best Model on Training

```python
svm_poly_Grid.fit(X_train, y_train)
best_model = svm_poly_Grid.best_estimator_
best_model
```

Evaluate on Testing

```python
evaluate_model(best_model, X_test, y_test)
```
